# 02 Build Preference Dataset

Converte o corpus original em dois datasets. DPO usa somente preferências inequívocas. O judge preserva A, B e empate.

```python
import sys
sys.path.append("../src")
import pandas as pd
from credit_explain.data.preference import to_dpo_pairs, to_judge_rows, deterministic_split

df = pd.read_csv("../data/raw/train.csv")
dpo = deterministic_split(to_dpo_pairs(df, include_swap=True))
judge = deterministic_split(to_judge_rows(df, include_swap=True))
dpo.shape, judge.shape
```

```python
dpo.split.value_counts(normalize=True), judge.groupby("split").label.value_counts(normalize=True)
```

A ordem das respostas é tratada como nuisance variable. Swap augmentation cria um teste direto de position bias.
